# Mission 5.3 — ReAct Tool-Calling Agent Loop
Run in Google Colab, inspect the Thought -> Action -> Observation trace, and upload the executed `.ipynb`.


```python
def tool_calc_roi(cost: float, benefit: float) -> str:
    roi = ((benefit - cost) / cost) * 100
    return f"ROI={roi:.1f}%"

def tool_policy_lookup(topic: str) -> str:
    kb = {"gpu": "Use T4 float16/bfloat16 for QLoRA fine-tuning.", "rag": "Chunk 400-800 chars with 15% overlap."}
    return kb.get(topic.lower(), "No internal record found.")

TOOLS = {"calc_roi": tool_calc_roi, "policy_lookup": tool_policy_lookup}
print("Registered tools:", list(TOOLS.keys()))
```

```python
MAX_STEPS = 4

def run_react_agent(goal: str):
    observations = []
    plan = [("calc_roi", (50, 140)), ("policy_lookup", ("rag",))]
    for step in range(1, MAX_STEPS + 1):
        if step - 1 >= len(plan):
            ans = " | ".join(observations)
            print(f"Final Answer: {ans}")
            return ans
        tool_name, args = plan[step - 1]
        print(f"Thought [{step}]: Need {tool_name} for {goal}")
        print(f"Action [{step}]: {tool_name}{args}")
        obs = TOOLS[tool_name](*args)
        observations.append(obs)
        print(f"Observation [{step}]: {obs}")

run_react_agent("Evaluate RAG ROI & chunking policy")
```
