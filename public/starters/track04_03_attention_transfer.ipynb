```python
import math
import torch
import torch.nn as nn
import torch.nn.functional as F

torch.manual_seed(42)
seq_len, d_k = 4, 8
Q = torch.randn(seq_len, d_k)
K = torch.randn(seq_len, d_k)
V = torch.randn(seq_len, d_k)

scores = (Q @ K.transpose(-2, -1)) / math.sqrt(d_k)
attn_weights = F.softmax(scores, dim=-1)
context = attn_weights @ V
print("Attention Weights Matrix (4x4):")
print(attn_weights.detach().numpy().round(3))
```

```python
# Transfer Learning Simulation: freeze backbone & train classification head only
backbone = nn.Sequential(nn.Linear(64, 128), nn.GELU(), nn.Linear(128, 64))
for param in backbone.parameters():
    param.requires_grad = False
head = nn.Linear(64, 4)
model = nn.Sequential(backbone, head)
trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total = sum(p.numel() for p in model.parameters())
print(f"Trainable params: {trainable} / {total} ({100*trainable/total:.1f}%)")
```
