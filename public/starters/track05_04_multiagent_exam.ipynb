```python
from typing import TypedDict, List

class AgentState(TypedDict):
    task: str
    research_notes: List[str]
    draft_answer: str
    approved: bool
    iterations: int

def tool_calc_roi(cost: float, benefit: float) -> str:
    roi = ((benefit - cost) / cost) * 100
    return f"ROI={roi:.1f}% (Benefit {benefit} vs Cost {cost})"
print("AgentState schema ready")
```

```python
def researcher_node(state: AgentState) -> AgentState:
    print(f"[Iter {state['iterations']+1} | Researcher] Calling tool_calc_roi...")
    state["research_notes"].append(tool_calc_roi(50, 140))
    state["iterations"] += 1
    return state

def writer_node(state: AgentState) -> AgentState:
    print("[Writer] Drafting executive report from tool observations...")
    evidence = " | ".join(state["research_notes"])
    state["draft_answer"] = f"Executive Summary for {state['task']}: {evidence}."
    return state

def reviewer_node(state: AgentState) -> AgentState:
    print("[Reviewer] Verifying quantitative evidence...")
    if "%" in state["draft_answer"] and state["iterations"] >= 1:
        state["approved"] = True
        print("[Reviewer] STATUS: APPROVED!")
    else:
        state["approved"] = False
    return state

state: AgentState = {"task": "RAG Infrastructure ROI", "research_notes": [], "draft_answer": "", "approved": False, "iterations": 0}
MAX_STEPS = 4
while not state["approved"] and state["iterations"] < MAX_STEPS:
    state = researcher_node(state)
    state = writer_node(state)
    state = reviewer_node(state)
print("\nFINAL REPORT:", state["draft_answer"])
```
